# 🍽️ Zomato Restaurant Data Analysis

## 🎯 Objective

Analyze restaurant data using Pandas to understand:

- Restaurant ratings
- Online ordering behavior
- Table booking availability
- Pricing trends
- Restaurant categories
- Customer voting patterns

## 🛠️ Tools

- Python
- Pandas
- NumPy
- Jupyter Notebook

```python
import pandas as pd

# Load the dataset
df = pd.read_csv(r"C:\Users\yatharth\Downloads\Zomato-data-.csv")

# Drop unnecessary columns to keep the dataset clean
df_clean = df.drop(columns=['url', 'phone', 'address'], errors='ignore')

# Check remaining columns and missing value counts
print(df_clean.isnull().sum())
```

```python
## exploring the dataset

print("Columns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
## converting this type of values (4.1/5) into numerical ratings

df["rate"] = df["rate"].str.split("/").str[0]

df["rate"] = pd.to_numeric(df["rate"], errors="coerce")

print(df["rate"].head())
```

```python
## checking
print("Missing Ratings:", df["rate"].isnull().sum())
```

```python
## stastics
print("Average Rating:", round(df["rate"].mean(), 2))
print("Highest Rating:", df["rate"].max())
print("Lowest Rating:", df["rate"].min())
print("Average Votes:", round(df["votes"].mean(), 2))
print("Average Cost for Two:", round(
    df["approx_cost(for two people)"].mean(), 2
))
```

```python
#top rated restraurants
top_restaurants = df.sort_values(
    by="rate",
    ascending=False
)[["name", "rate", "votes"]].head(10)

top_restaurants
```

```python
## cost analysis
cost_analysis = df.groupby(
    "listed_in(type)"
)["approx_cost(for two people)"].mean().sort_values(
    ascending=False
)

cost_analysis
```

```python
## online order analysis
online_order_analysis = df.groupby(
    "online_order"
).agg(
    restaurants=("name", "count"),
    avg_rating=("rate", "mean"),
    avg_votes=("votes", "mean"),
    avg_cost=("approx_cost(for two people)", "mean")
).round(2)

online_order_analysis
```

```python
online_order_count = df["online_order"].value_counts()

online_order_count
```

```python
## table booking analysis

booking_analysis = df.groupby(
    "book_table"
).agg(
    restaurants=("name", "count"),
    avg_rating=("rate", "mean"),
    avg_votes=("votes", "mean"),
    avg_cost=("approx_cost(for two people)", "mean")
).round(2)

booking_analysis
```

```python
## Most Popular Restaurants
popular_restaurants = df.sort_values(
    by="votes",
    ascending=False
)[["name", "rate", "votes", "online_order"]].head(10)

popular_restaurants
```

```python
## Find High-Rated & Affordable Restaurants
value_restaurants = df[
    (df["rate"] >= 4.0) &
    (df["approx_cost(for two people)"] <= 500)
].sort_values(
    by=["rate", "votes"],
    ascending=[False, False]
)

value_restaurants[
    ["name", "rate", "votes", "approx_cost(for two people)"]
].head(10)
```

## what the notebook  taught and conveyed to us

- Analyzed the relationship between online ordering and restaurant performance.
- Compared restaurants with and without table-booking facilities.
