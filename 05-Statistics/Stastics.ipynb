#  Descriptive Statistics for Data Science

A practical exploration of descriptive statistics using Python.

##  Objectives

- Understand central tendency
- Measure data dispersion
- Analyze percentiles and quartiles
- Detect outliers
- Calculate Z-scores
- Understand correlation
- Extract statistical insights

## Tools

Python • NumPy • Pandas • Matplotlib • Seaborn

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Study_Hours": [2, 3, 4, 5, 5, 6, 7, 8, 9, 10],
    "Marks": [52, 58, 65, 70, 72, 78, 82, 88, 91, 95]
}

df = pd.DataFrame(data)

df
```

```python
marks = df["Marks"]

print("Mean:", marks.mean())
print("Median:", marks.median())
print("Mode:", marks.mode().tolist())
```

###  Interpretation

- **Mean** represents the average score.
- **Median** represents the middle value.
- **Mode** represents the most frequently occurring value.

These measures help describe the central tendency of a dataset.

```python
data_range = marks.max() - marks.min()
variance = marks.var()
std_dev = marks.std()

print("Range:", data_range)
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(std_dev, 2))
```

```python
Q1 = marks.quantile(0.25)
Q2 = marks.quantile(0.50)
Q3 = marks.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Median (Q2):", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Marks"] < lower_limit) |
    (df["Marks"] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nOutliers:")
print(outliers if not outliers.empty else "No outliers detected")
```

```python
df["Z_Score"] = (
    (df["Marks"] - df["Marks"].mean())
    / df["Marks"].std()
)

df[["Student", "Marks", "Z_Score"]]
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="Marks",
    kde=True
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["Marks"])

plt.title("Marks Distribution and Outliers")
plt.xlabel("Marks")

plt.show()
```

```python
correlation = df["Study_Hours"].corr(df["Marks"])

print("Correlation between Study Hours and Marks:",
      round(correlation, 2))
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
summary = df[["Study_Hours", "Marks"]].describe()

summary
```

#  Key Insights

This analysis demonstrated:

- Measures of central tendency using mean, median and mode.
- Measures of dispersion using variance and standard deviation.
- Quartiles and IQR for understanding data spread.
- IQR-based outlier detection.
- Z-score calculation for standardized values.
- Correlation analysis between study hours and marks.
