#  Seaborn Fundamentals

A practical introduction to Seaborn for statistical data visualization.

## Objectives

- Create attractive statistical visualizations
- Understand categorical and numerical relationships
- Analyze distributions
- Compare groups
- Visualize correlations
- Customize plots

## Libraries

- Python
- Pandas
- NumPy
- Seaborn
- Matplotlib

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

print("Seaborn version:", sns.__version__)
```

```python
df = sns.load_dataset("tips")

print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nBasic Statistics:")
df.describe()
```

```python
## Relational visualization(Scatter PLot)
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip"
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

```python
##Scatter plot with usage with hue
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time"
)

plt.title("Total Bill vs Tip by Time")

plt.show()
```

```python
## Categorial Visualization(count plot)
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="day"
)

plt.title("Number of Records by Day")
plt.xlabel("Day")
plt.ylabel("Count")

plt.show()
```

```python
## Barplot
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="day",
    y="total_bill",
    errorbar=None
)

plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Bill")

plt.show()
```

```python
## boxplot 
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.title("Total Bill Distribution by Day")

plt.show()
```

```python
## violin plot
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.title("Total Bill Distribution by Day")

plt.show()
```

```python
## histogram
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="total_bill",
    bins=20,
    kde=True
)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

```python
## distribution about gender
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="total_bill",
    hue="sex",
    kde=True,
    element="step"
)

plt.title("Total Bill Distribution by Gender")

plt.show()
```

```python
## correlaton matrix **
numeric_data = df.select_dtypes(include=np.number)

correlation = numeric_data.corr()

correlation

```

```python
## heatmap for matrix
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
## A summarized plot 
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="day",
    size="size",
    alpha=0.8
)

plt.title("Customer Bill vs Tip Analysis", fontsize=16)
plt.xlabel("Total Bill", fontsize=12)
plt.ylabel("Tip", fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```
